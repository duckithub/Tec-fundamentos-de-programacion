### Ejercicio 3: Contador de dígitos
**Enunciado:** Crea un programa que pida un número entero positivo y cuántos
dígitos tiene, usando `while` con el operador división entera `//` (cada iteración
elimina el último dígito).
**Entrada:** `4572`
**Proceso:** `4572 -> 457 -> 45 -> 4 -> 0` (4 eliminaciones)
**Salida:** `El número 4572 tiene 4 dígitos`
---

````python
num = int(input("numero: "))
numoriginal = num
contador = 0 

while num > 0:
    num = num // 10
    contador = contador + 1

print("el numero", numoriginal, "tiene", contador, "digitos")     
          
````

### Ejercicio 4: Suma de dígitos
**Enunciado:** Crea un programa que pida un número entero positivo y calcule la
suma de sus dígitos usando `while`, el módulo `%` para extraer el último dígito y
`//` para eliminarlo.
**Entrada:** `4729`
**Proceso:** `4 + 7 + 2 + 9 = 22`
**Salida:** `La suma de los dígitos de 4729 es: 22`
---

````python
num = int(input("Ingrese un numero: "))
numoriginal = num 
suma = 0 
while num > 0: 
    digito = num %10
    suma = suma + digito
    num = num// 10
print("la suma de digitos de", numoriginal, "es", suma)    
````

### Ejercicio 5: Número invertido
**Enunciado:** Crea un programa que pida un número entero positivo y muestre sus
dígitos en orden inverso. Construye el nuevo número en un acumulador con la fórmula
`invertido = invertido * 10 + digito`.
**Entrada:** `1234`
**Proceso:** `4`, luego `43`, luego `432`, luego `4321`
**Salida:** `El número invertido de 1234 es 4321`
---

````python
num = int(input("introduzca numero: "))
invertido = 0 
while num != 0:
    r = num %10
    invertido = invertido * 10 + r
    num = num //10 

print(f"{invertido}")   
````
