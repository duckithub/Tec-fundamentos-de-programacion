### Extra 1: Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.
**Entrada:**
```
Nueva calificación 1: 8.5
Nueva calificación 2: 9.5
```
**Salida:**
```
Tercera calificación: 8.0
Nueva tupla: (7.5, 9.0, 8.0, 6.5, 10.0, 8.5, 9.5)
Lista ordenada (mayor a menor): [10.0, 9.5, 9.0, 8.5, 8.0, 7.5, 6.5]
Suma total: 59.0
```

````python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

calif5 = 0
calif6 = 0

calif5 = float(input("Ingrese la calificación 5: "))
calif6 = float(input("Ingrese la calificación 6: "))

nuevas_calificaciones = calificaciones + (calif5, calif6)

calificaciones = list(calificaciones)
calificaciones.sort(reverse=True)

print("Aqui estan las calificaciones: ")
print(calificaciones)

print("\nEstas son las calificaciones actualizadas: ")
print(nuevas_calificaciones)

print("\nEstas son las calificaciones finales de mayor a menor: ")
print(sorted(nuevas_calificaciones, reverse=True))
````

### Extra 2: Agenda de contactos con búsqueda
**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.
**Entrada:**
```
Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro
```
**Salida:**
```
Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140
```

````python
agenda = {"Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103" }

nombre =  input("Ingrese el nombre del nuevo contacto:")
telefono = input("ingrese el telefono del nuevo contacto: ")
agenda[nombre] = telefono

def buscar_contacto(agenda, nombre):
    return agenda.get(nombre, "No encontrado")

nombre = input("Ingrese el nombre del contacto a buscar: ")

print("Contanctos registrados: ")
for n in agenda.keys():
    print(n)
    
print("\nEl teléfono de", nombre, "es:", buscar_contacto(agenda, nombre))
````

### Extra 3: Calculadora segura con manejo de excepciones
**Enunciado:** Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.
**Entrada:**
```
Primer número entero: 20
Segundo número entero: 4
```
**Salida:**
```
20 dividido entre 4 es: 5.0
```
**Caso de error (divisor cero):**
```
Primer número entero: 10
Segundo número entero: 0
```
**Salida:**
```
Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.
```

````python
try:
    dividiendo = int(input("ingrese un numero: "))
    divisor = int(input("ingrese otro numero: "))
    resultado = dividiendo / divisor
    print(f"El resultado de la división de {dividiendo} / {divisor} es: {resultado}")
except ValueError:
    print("Error: Por favor, ingrese números válidos.")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
````

### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

````python
votos = {"ROJO": 0, "AZUL": 0, "VERDE": 0}

cantidad = int(input("cuantos votos van a ser?: (MAX 5)"))

conteo = 0
while conteo < cantidad:
    conteo += 1
    voto = input(f"ingrese el voto {conteo} (Rojo, Azul, Verde): ")
    print(f"votaste por {voto.upper()}")

    if voto.upper() in votos:
        votos[voto.upper()] += 1
    else:
        print("voto invalido")

print("\nResultados: ")
for k,v in votos.items():
    print(f"el color {k} tiene {v} votos")

color_ganador = None
votos_ganador = -1

for color, conteo_votos in votos.items():
    if conteo_votos > votos_ganador:
        votos_ganador = conteo_votos
        color_ganador = color

print(f"\nEl color ganador es {k} con {v} votos.")

````
