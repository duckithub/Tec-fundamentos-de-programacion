```python
def suma_tupla(t):
    total = 0 
    for e in t:
        total += e
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)
print(f"El total de calorias es: {resultado}")
```

```python
print("impresion de llaves")
for n in contactos.keys():
    print(n)

print("\nimpresion de valores")
for v in contactos.values():
    print(v)

print("\nimpresion completa")
for n, v in contactos.items():
    print(f"nombres: {n}, telefono: {v}")
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "No encontrado")

nombre = input("Ingrese el nombre del contacto a buscar: ")
telefono = buscar_contacto(contactos, nombre)
print (f"El telefono de {nombre} es: {telefono}")
```

### EXEPCIONES



```python
try:
    edad = int(input("ingrese su  edad:"))
    print(f"Su edad es: {edad}")
except ValueError:
    print("Error: Por favor, ingrese un número válido para la edad.")
```

```python
try:
    dividiendo = int(input("ingrese un numero: "))
    divisor = int(input("ingrese otro numero: "))
    resultado = dividiendo / divisor
    print(f"El resultado de la división de {dividiendo} / {divisor} es: {resultado}")
except ValueError:
    print("Error: Por favor, ingrese números válidos.")
except ZeroDivisionError:
    print("Error: El divisor no puede ser cero.")
finally:
    print("ciao")
```

### EJERCICIOS


```python
candidatos = {
    "A":0,
    "B":0,
    "C":0
}

cantidad = int(input("cuantos votos van a ser?: "))

conteo = 1
while conteo <= cantidad:
    voto = input(f"ingrese el voto {conteo} (A, B, C): ")
    print(f"votaste por {voto.upper()}")
   

if voto.upper() in candidatos:
    candidatos[voto.upper()] += 1
    conteo += 1
else:
    print("voto invalido")

print("\n resultados")
for k,v in candidatos.items():
    print(f"candidato {k} tiene {v} votos")
```
