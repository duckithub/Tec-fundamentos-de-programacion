```python
#Ejercicio 5

print("Que tipo de visitante eres?: regular, estudiante, profesor, adulto mayor")
visitante = input("Tipo de visitante: ")

precio = 45

dis =0
dis_est =0.10
dis_prof =0.10
dis_adul =0.12

if visitante == "regular":
    float(dis == 0)
elif visitante == "estudiante":
    float(dis_est == 0.10 )
elif visitante == "profesor":
    float(dis_prof == 0.10 )
elif visitante == "adulto mayor":
    float(dis_adul == 0.12 )
else:
    print("Tipo de visitante no válido.")

precio_final = precio - (precio * (dis or dis_est or dis_prof or dis_adul) / 100)

print("El precio final es:", precio_final)

```
