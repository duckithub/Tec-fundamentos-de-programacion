```python
def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total 
numeros = [10, 20, 30, 40]


resultado = sumar_lista(numeros)

print(f"La suma de los elementos es: {resultado}")
```

```python
def encontrar_maximo(lista):
    maximo = lista[0]
    
    for elemento in lista:
        if elemento > maximo:
            maximo = elemento
            
    return maximo

numeros = [7, 3, 9, 2]

resultado = encontrar_maximo(numeros)

print(f"El elemento máximo es: {resultado}")
```

```python
def sumar_diagonal_principal(matriz):
    suma = 0
    
    for i in range(len(matriz)):
        suma += matriz[i][i]
    return suma

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

resultado = sumar_diagonal_principal(matriz)

print(f"La suma de la diagonal principal es: {resultado}")
```

```python
def calcular_y_mostrar_promedio(lista):
    
    if len(lista) == 0:
        print("La lista está vacía.")
        return
    
    suma = 0
    for elemento in lista:
        suma += elemento

    promedio = suma / len(lista)

    print(f"El promedio es: {promedio:.2f}")

numeros = [8, 9, 10, 7]

calcular_y_mostrar_promedio(numeros)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

numero = 5

resultado = factorial(numero)

print(f"El factorial de {numero} es: {resultado}")
```

```python
def procesar_tupla(datos):
    tercer_elemento = datos[2]

    suma_total = 0
    for elemento in datos:
        suma_total += elemento
        
    return tercer_elemento, suma_total

numeros = (4, 8, 15, 16, 23, 42)

tercer_num, suma = procesar_tupla(numeros)

print(f"El tercer elemento es: {tercer_num}")
print(f"La suma de la tupla es: {suma}")
```

```python
def buscar_telefono(directorio, nombre):
    if nombre in directorio:
        return f"El teléfono de {nombre} es: {directorio[nombre]}"
    else:
        return f"El contacto '{nombre}' no fue encontrado."

agenda = {
    "Ana": "5551234",
    "Beto": "5555678",
    "Carlos": "5559012"
}

contacto = "Beto"

resultado = buscar_telefono(agenda, contacto)
print(resultado)
```

```python
def calcular_promedios(alumnos):
    for nombre, calificaciones in alumnos.items():
        promedio = sum(calificaciones) / len(calificaciones)
        print(f"{nombre}: {promedio:.2f}")

calificaciones_alumnos = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

calcular_promedios(calificaciones_alumnos)
```

```python
def dividir(a, b):
    try:
        resultado = a / b
        return f"El resultado es: {resultado}"
    except ZeroDivisionError:
        return "Error: No se puede dividir entre cero."

num1 = 10
num2 = 0

print(dividir(num1, num2))
```

```python
def pedir_entero(entrada):
    try:
        numero = int(entrada)
        return f"Número ingresado: {numero}"
    except ValueError:
        return "Error: Debes ingresar un número entero."

valor = "abc"

print(pedir_entero(valor))
```

```python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)

mensaje = "Python es un lenguaje de programacion"

total = contar_palabras(mensaje)

print(f"El mensaje tiene {total} palabras.")
```

```python
def procesar_texto(texto):
    
    longitud = len(texto)
    
    mayusculas = texto.upper()
    
    reemplazado = texto.replace("python", "código")
    
    return longitud, mayusculas, reemplazado

entrada = "aprende python"

longitud, mayusculas, reemplazado = procesar_texto(entrada)

print(f"Longitud: {longitud}")
print(f"Mayúsculas: {mayusculas}")
print(f"Reemplazo: {reemplazado}")
```
