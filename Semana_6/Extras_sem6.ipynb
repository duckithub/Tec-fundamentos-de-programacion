```python
base = float(input("Dame la base: "))
altura = float(input("Dame la altura: "))
area = (base * altura) / 2
print("El area es de:", area)
```

```python
c = float(input("Dame la temperatura en Celsius"))
f = c * 9/5 + 32
print(c,"grados celcius equivalen a:",f,"grados Farenheit")
```

```python
capital = float(input("Dame la capital: "))
tasa = float(input("Dame la tasa: "))
tiempo = float(input("Dame el tiempo: "))
interes = capital * (tasa / 100) * tiempo

print("El interes simple es:",interes )
```
